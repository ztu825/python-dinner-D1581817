## 題目說明

朋友聚餐的餐費是含貨幣符號與逗號的文字,人數也是文字,需先整理資料、轉換型別再計算。

規則:
1. 服務費 = 原始餐費 × 10%
2. 應付總額 = 原始餐費 + 服務費 − 折價券
3. 每人費用 = 應付總額 ÷ 人數
4. 每人預算 300 元,用比較運算判斷是否超過

每組測試都會重新執行完整的整理、計算與輸出程式。

## 測試一:原始資料(4 人、折價券 100 元)

預期:應付總額 1,220、每人 305、超過預算 `True`

```python
# ===== 原始資料(只修改本組測試指定的項目)=====
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# ===== 1. 整理聚餐名稱:去頭尾空白、轉小寫 =====
clean_event_name = event_name.strip().lower()

# ===== 2. 整理餐費並轉換型別(保留原始資料,結果存到新變數)=====
# 移除 "NT$"、逗號與空白,再轉成可計算的浮點數
meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_value = float(meal_text)
# 文字人數 "4" 轉成整數 4,才能拿來做除法
people_count = int(people)

# ===== 3. 分步計算費用 =====
service_fee = meal_value * service_rate            # 服務費:以「原始餐費」計算
total_payable = meal_value + service_fee - discount  # 應付總額:餐費 + 服務費 - 折價券
cost_per_person = total_payable / people_count     # 每人費用

# ===== 4. 判斷是否超過預算(比較運算,結果為布林值)=====
over_budget = cost_per_person > budget_per_person

# ===== 5. 輸出費用摘要 =====
print(f"聚餐名稱:{clean_event_name}")
print(f"原始餐費:{meal_value:.2f} 元")
print(f"服務費:{service_fee:.2f} 元")
print(f"折價券:{discount:.2f} 元")
print(f"應付總額:{total_payable:.2f} 元")
print(f"分攤人數:{people_count} 人")
print(f"每人費用:{cost_per_person:.2f} 元")
print(f"每人預算:{budget_per_person:.2f} 元")
print(f"是否超過預算:{over_budget}")
```
